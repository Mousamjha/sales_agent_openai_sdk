### Streaming text

```python
result = Runner.run_streamed(agent, "Tell me a brain tickling joke")
async for event in result.stream_events():
    if event.type == "raw_response_event" and isinstance(event.data, ResponseTextDeltaEvent):
        print(event.data.delta, end="", flush=True)
```

### Adding tool

```python
PUSHOVER_USER = os.getenv("PUSH_USER_KEY")
PUSHOVER_TOKEN = os.getenv("PUSH_API_TOKEN")
PUSHOVER_URL = "https://api.pushover.net/1/messages.json"
if PUSHOVER_TOKEN:
    if PUSHOVER_TOKEN.startswith('a'):
        print("Pushover Token Configured")
    else:
        print("Pushover token not configured")
else:
    print("Pushover user token not found")
if PUSHOVER_USER:
    if PUSHOVER_USER.startswith("u"):
        print("Pushover user configured")
    else:
        print("Pushover user not configured")
else:
    print("Pushover user not fount")
```

```python
def push(message):
    print(f"Push: {message}")
    payload = {
        "user": PUSHOVER_USER,
        "token": PUSHOVER_TOKEN,
        "message": message 
    }
    requests.post(PUSHOVER_URL, data=payload)
```

```python
push("Hey")
```

```python
@function_tool
def push_tool(message: str) -> str:
    """Send the message to the user as push notification"""
    print(f"Push: {message}")
    payload = {
        "user": PUSHOVER_USER,
        "token": PUSHOVER_TOKEN,
        "message": message
    }
    result = requests.post(url = PUSHOVER_URL, data=payload)
    return f"PUSH sent with api status code {result}"
```

```python
push_tool.description
```

```python
notifier = Agent(
        name="Notifier", 
        model='gpt-5.4-nano', 
        instructions="You notify the user when requested", 
        tools=[push_tool]
        )
```

```python
with trace("Notifying the user"):
    result = await Runner.run(notifier, f"Notfiy the user with jokes")
print(result)
print(result.final_output)
```

### Session Memory (short-term-memory)

```python
agent = Agent(name="Assistant", model='gpt-5.4-nano')
```

```python
response = await Runner.run(agent, "Hi there my name is Mousam")
print(response.to_input_list())
```

```python
response = await Runner.run(agent, "What is my name")
response
```

```python
res = await Runner.run(agent, "Hi, my name is Mousam")
input = res.to_input_list() + [{"role":"user","content": "What is my name?"}]
nextPrompt = await Runner.run(agent, input)
print(nextPrompt.final_output)
```

```python
session = SQLiteSession("12345")
```

```python
response = await Runner.run(agent, "Hi there, my name is Mousam", session=session)
```
